Dataset Generation for First Project "ACADEMIC PERFORMANCE INSIGHTS"

```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import random

# Define parameters
num_of_stdts=42
std_first_names=["Aarav", "Aditi", "Arjun", "Ananya", "Bhavya", "Chaitanya", "Deepak", "Divya", "Ishani", "Karan", "Lakshmi", "Meera", "Nikhil", "Pooja", "Rahul", "Sneha"]
std_last_names=["Sharma", "Reddy", "Iyer", "Gupta", "Nair", "Rallabandi", "Verma", "Joshi", "Malhotra", "Kulkarni"]
Gender=['Male','Female']

# Generating 42 Student records with no duplicates or missing values
uni_data = {
    'Rno': [f"COL{100+i}" for i in range(1,num_of_stdts+1)],
    'Name': [f"{random.choice(std_first_names)} {random.choice(std_last_names)}" for i in range(1,num_of_stdts+1)],
    'Gender': [random.choice(Gender) for i in range(num_of_stdts)],
    'Maths': [random.randint(1,100) for i in range(num_of_stdts)],
    'Physics': [random.randint(1,100) for i in range(num_of_stdts)],
    'Chemistry': [random.randint(1,100) for i in range(num_of_stdts)]
}

# Converting the 42 records into a Data frame
uni_df=pd.DataFrame(uni_data)

# generating 5 student records with duplicate values
dup_df=uni_df.sample(n=5, random_state=42)

# Generating 3 student records with Missing values
null_clmns=['Name','Gender','Maths','Physics','Chemistry']

null_rcds=[]

for i in range(3):
    rcd={
        'Rno': f"COL{143+i}",
        'Name': f"{random.choice(std_first_names)} {random.choice(std_last_names)}",
        'Gender': random.choice(Gender),
        'Maths': random.randint(1,100),
        'Physics': random.randint(1,100),
        'Chemistry': random.randint(1,100)
    }

    random_col=random.choice(null_clmns)

    rcd[random_col]=None

    null_rcds.append(rcd)

# Convert the generated records with missing values into a Data frame
null_df=pd.DataFrame(null_rcds)

# Combining all three Data frames (Unique, Duplicate and Missing Value Records) into a single final Data frame
final_df=pd.concat([uni_df,dup_df,null_df], ignore_index=True)

# Shuffle the records so that they are jumbled and mixed up
final_df = final_df.sample(frac=1).reset_index(drop=True)

# Saving Dataframe as .csv file in the same folder as this notebook
final_df.to_csv('Student_info.csv',index=False)

print("Dataset is created successfully and is saved to 'Student_info.csv' in the current Folder!!")
```

Step 2: Data Transformation

```python
#Add a new column Total Marks = S1 + S2 + S3
std_df['Total Marks']=std_df['Maths']+std_df['Physics']+std_df['Chemistry']
print(std_df)
```

```python
#Display only female students with their Total Marks
result6=std_df.loc[std_df['Gender']=='Female',['Rno','Total Marks']]
print(result6)
```

```python
#Calculate percentage for each student and add a Percentage column
std_df['Percentage']=((std_df['Maths']+std_df['Physics']+std_df['Chemistry'])/300.0)*100
std_df['Percentage']=std_df['Percentage'].round(2)
std_df
```

```python
'''
Add a new column Grade based on the following conditions:
-> A: 80% – 100%
-> B: 60% – 79%
-> C: 40% – 59%
-> FAIL: Less than 40%
'''
def assign_grade(pt):
    if pd.isna(pt):
        return "Invalid Percentage Value"
    elif pt>=80 and pt<=100:
        return "A"
    elif pt>=60 and pt<=79:
        return "B"
    elif pt>=40 and pt<=59:
        return "C"
    else:
        return "FAIL"
std_df['Grade']=std_df['Percentage'].apply(assign_grade)
std_df
```

Step 3: Data Visualization

```python
#Create a Bar Chart showing Percentage of Male vs Female students
from matplotlib import pyplot as plt
import pandas as pd
grouped_by_gen=std_df.groupby('Gender')
groupby_obj=grouped_by_gen['Percentage']
avg_genders=groupby_obj.mean()
plt.bar(avg_genders.index,avg_genders.values,color=['blue','red'])
plt.title("Percentage of Males vs Females")
plt.xlabel("Gender")
plt.ylabel("Percentage")
plt.show()
```

```python
# Create a Pie Chart showing the PASS vs FAIL percentage of all students
from matplotlib import pyplot as plt
import pandas as pd
pass_stds=std_df[std_df['Grade'].isin(['A','B','C'])]
fail_stds=std_df[std_df['Grade']=="FAIL"]
num_pass_stds=len(pass_stds)
num_fail_stds=len(fail_stds)
plt.title("Percentage of Passed VS Failed Students")
plt.pie([num_pass_stds, num_fail_stds], labels=['PASS', 'FAIL'], autopct="%0.1f%%",explode=[0,0],wedgeprops={'edgecolor':'yellow','linewidth':1},colors=["blue","red"])
plt.show()
```
